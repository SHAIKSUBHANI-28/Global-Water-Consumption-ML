```python
# Identify categorical and numerical columns

categorical_columns = df.select_dtypes(include=["object"]).columns.tolist()
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

print("Categorical columns:")
print(categorical_columns)

print("\nNumerical columns:")
print(numerical_columns)
#I identified the categorical and numerical columns based on their data types. Country and Water Scarcity Level were identified as categorical variables, while Year and the seven water and environmental measurements were identified as numerical variables. This helps determine which preprocessing technique should be applied to each type of variable.
```

```python
# One-hot encode the Country column

df_encoded = pd.get_dummies(
    df.drop(columns=["Water Scarcity Level"]),
    columns=["Country"],
    dtype=int
)

print("Dataset shape after encoding:", df_encoded.shape)

print("\nFirst 5 rows after encoding:")
display(df_encoded.head())
#In this step, I removed the Water Scarcity Level column because it is not the target of this regression problem. I then applied One-Hot Encoding to the Country column because it is a categorical input variable. The 20 countries were converted into 20 binary columns containing 0 and 1. As a result, the dataset changed from 10 original columns to 28 encoded columns while retaining all 500 observations.
#i removed Water Scarcity Level,Because my approved target variable is Per Capita Water Use, and Water Scarcity Level is not required as an input feature for this Linear Regression model.
```

```python
# Separate features and target

X = df_encoded.drop(
    "Per Capita Water Use (Liters per Day)",
    axis=1
)

y = df_encoded["Per Capita Water Use (Liters per Day)"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())
#In this step, I separated the dataset into input features X and target variable y. I removed Per Capita Water Use from the feature set because it is the value we want to predict. The remaining 27 columns, including seven numerical features and 20 one-hot encoded country features, are used as inputs. Therefore, X contains 500 observations and 27 features, while y contains 500 target values.
```

```python
# Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
#I split the dataset into 80% training and 20% testing data using train_test_split. From the 500 observations, 400 were used to train the Linear Regression model and 100 were reserved for testing. The 27 input features are present in both sets, while the target is Per Capita Water Use. I used random_state 42 to make the split reproducible.
```

```python
# Feature scaling

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully!")
print("Scaled training features shape:", X_train_scaled.shape)
print("Scaled testing features shape:", X_test_scaled.shape)
#I applied feature scaling to the 27 input features using StandardScaler. The scaler was fitted only on the training data and then used to transform both the training and testing data. This puts the features on a comparable scale and prevents data leakage. The target variable, Per Capita Water Use, was not scaled.
```

```python
# Build the Linear Regression model

linear_model = LinearRegression()

linear_model.fit(X_train_scaled, y_train)

print("Linear Regression model trained successfully!")
#In this step, I created a Linear Regression model and trained it using the 400 training observations and 27 scaled input features. The target variable is Per Capita Water Use in liters per day. The model learns the linear relationship between the input features and the target using the training data.
```

```python
# Make predictions on the test data

y_pred = linear_model.predict(X_test_scaled)

print("Predictions generated successfully!")
print("\nFirst 10 actual values:")
print(y_test.head(10).values)

print("\nFirst 10 predicted values:")
print(y_pred[:10])
#After training the Linear Regression model, I used the 100 unseen test observations to generate predictions. The model used the 27 scaled input features to predict Per Capita Water Use. I then displayed the first 10 actual values and their corresponding predicted values for comparison.
```

```python
# Evaluate the Linear Regression model

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Linear Regression Performance")
print("-----------------------------")
print("MAE  :", round(mae, 4))
print("MSE  :", round(mse, 4))
print("RMSE :", round(rmse, 4))
print("R²   :", round(r2, 4))
#MAE tells us the average absolute difference between the actual and predicted water usage.
#I evaluated the Linear Regression model by comparing the actual test values with the predicted values. I calculated MAE, MSE, RMSE, and R². MAE, MSE, and RMSE measure prediction error, where lower values are better, while R² measures how well the model explains the variation in Per Capita Water Use.
```

```python
# Actual vs Predicted values

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Per Capita Water Use (Liters per Day)")
plt.ylabel("Predicted Per Capita Water Use (Liters per Day)")
plt.title("Actual vs Predicted Water Usage")

# Perfect prediction line
min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.show()
#I created an Actual vs Predicted scatter plot to visually evaluate the Linear Regression model. The X-axis represents the actual Per Capita Water Use values, while the Y-axis represents the predicted values. The dashed diagonal line represents perfect prediction. Points closer to this line indicate more accurate predictions, while points farther away indicate larger errors.
#Each dot represents one test observation, comparing its actual water usage with the value predicted by the model.
#The dashed line represents perfect prediction, where the actual value equals the predicted value
```

```python
# Detect outliers using the IQR method

numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns

print("Outlier count for each numerical feature:")

for column in numeric_features:
    Q1 = X[column].quantile(0.25)
    Q3 = X[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = X[
        (X[column] < lower_limit) |
        (X[column] > upper_limit)
    ]

    print(column, "→", len(outliers), "outliers")
    #I detected potential outliers in the numerical input features using the Interquartile Range method. For each numerical feature, I calculated Q1, Q3, and IQR, and values below Q1 minus 1.5 times IQR or above Q3 plus 1.5 times IQR were identified as potential outliers. At this stage, I only detected the outliers; I did not remove them.
    #Some extreme values may represent genuine water-consumption or environmental conditions, so instead of deleting observations, I handle them using IQR capping in the next preprocessing step.
```

```python
# Handle numerical outliers using IQR capping

X_train_capped = X_train.copy()
X_test_capped = X_test.copy()

numeric_features = [
    "Year",
    "Total Water Consumption (Billion Cubic Meters)",
    "Agricultural Water Use (%)",
    "Industrial Water Use (%)",
    "Household Water Use (%)",
    "Rainfall Impact (Annual Precipitation in mm)",
    "Groundwater Depletion Rate (%)"
]

for column in numeric_features:
    Q1 = X_train_capped[column].quantile(0.25)
    Q3 = X_train_capped[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    X_train_capped[column] = X_train_capped[column].clip(
        lower_limit,
        upper_limit
    )

    X_test_capped[column] = X_test_capped[column].clip(
        lower_limit,
        upper_limit
    )

print("Numerical outliers handled using IQR capping.")
print("Training data shape:", X_train_capped.shape)
print("Testing data shape:", X_test_capped.shape)
#I handled numerical outliers using IQR capping. I used seven continuous numerical input features and calculated Q1, Q3, and IQR using only the training data. Values below the lower IQR limit or above the upper IQR limit were capped at those limits instead of being removed. The same limits calculated from the training data were applied to the test data to avoid data leakage.
#I used capping because extreme values may represent genuine environmental conditions. Capping reduces their influence without removing the observations.
#I handled outliers in seven continuous numerical features: Year, Total Water Consumption, Agricultural Water Use, Industrial Water Use, Household Water Use, Rainfall Impact, and Groundwater Depletion Rate.
```

```python
# Scale the outlier-handled data

scaler = StandardScaler()

X_train_final = scaler.fit_transform(X_train_capped)
X_test_final = scaler.transform(X_test_capped)

print("Feature scaling completed after outlier handling.")
print("Final training features shape:", X_train_final.shape)
print("Final testing features shape:", X_test_final.shape)
#After handling the numerical outliers using IQR capping, I performed feature scaling again on the modified data using StandardScaler. The scaler was fitted only on the training data and then applied to the test data. This produced the final scaled training and testing feature sets, containing 400 and 100 observations respectively, with 27 input features.
#I scale again ,Because the outlier-capping step changed some feature values. Therefore, I scaled the final outlier-handled data before training the final Linear Regression model.
```

```python
# Train the final Linear Regression model

linear_model_final = LinearRegression()

linear_model_final.fit(
    X_train_final,
    y_train
)

print("Final Linear Regression model trained successfully!")
#After handling outliers and scaling the features, I trained the final Linear Regression model using the 400 training observations and 27 input features. The target variable was Per Capita Water Use in liters per day. This final model uses the preprocessed data after outlier capping and feature scaling.
#The first model was an initial model trained before outlier handling. After detecting and capping the outliers, I created the final preprocessed training data and trained a second final Linear Regression model. The final model is used for the final evaluation.
```

```python
# Make predictions using the final model

y_pred_final = linear_model_final.predict(X_test_final)

print("Final predictions generated successfully!")

print("\nFirst 10 actual values:")
print(y_test.head(10).values)

print("\nFirst 10 predicted values:")
print(y_pred_final[:10])
#After training the final Linear Regression model, I used the final scaled test features to generate predictions for the 100 unseen test observations. The predicted values are stored in y_pred_final. I also displayed the first 10 actual and predicted values so that they can be compared before calculating the final evaluation metrics.
#y_pred contains predictions from the initial model before outlier handling, while y_pred_final contains predictions from the final model after outlier handling and feature scaling
```

```python
# Evaluate the final Linear Regression model

final_mae = mean_absolute_error(y_test, y_pred_final)
final_mse = mean_squared_error(y_test, y_pred_final)
final_rmse = np.sqrt(final_mse)
final_r2 = r2_score(y_test, y_pred_final)

print("Final Linear Regression Performance")
print("-----------------------------------")
print("MAE  :", round(final_mae, 4))
print("MSE  :", round(final_mse, 4))
print("RMSE :", round(final_rmse, 4))
print("R²   :", round(final_r2, 4))
#I evaluated the final Linear Regression model using the 100 unseen test observations. I compared the actual Per Capita Water Use values with the final predicted values and calculated MAE, MSE, RMSE, and R². MAE, MSE, and RMSE measure prediction error, while R² measures how well the model explains the variation in the target
#these called final metrics,Because these metrics are calculated using predictions from the final model trained after IQR outlier capping and feature scaling.
```

```python
# Final Actual vs Predicted graph

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred_final)

plt.xlabel("Actual Per Capita Water Use (Liters per Day)")
plt.ylabel("Predicted Per Capita Water Use (Liters per Day)")
plt.title("Final Linear Regression: Actual vs Predicted")

# Perfect prediction line
min_value = min(y_test.min(), y_pred_final.min())
max_value = max(y_test.max(), y_pred_final.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.show()
#I created the final Actual vs Predicted scatter plot using the predictions from the final Linear Regression model. The X-axis represents the actual Per Capita Water Use values and the Y-axis represents the predicted values. The dashed diagonal line represents perfect prediction. The closer the points are to this line, the more accurate the predictions are.
#The earlier graph showed predictions from the initial model. This final graph shows predictions after IQR outlier capping and final feature scaling, so it represents the performance of my final model.
```

```python
# 5-Fold Cross-Validation for Linear Regression

from sklearn.model_selection import KFold, cross_val_score
from sklearn.pipeline import Pipeline

# Create a pipeline
linear_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LinearRegression())
])

# 5-fold cross-validation
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_r2_scores = cross_val_score(
    linear_pipeline,
    X,
    y,
    cv=cv,
    scoring="r2"
)

print("5-Fold Cross-Validation R² Scores:")
print(cv_r2_scores)

print("\nMean Cross-Validation R²:",
      round(cv_r2_scores.mean(), 4))
#I used 5-fold cross-validation to evaluate the consistency of the Linear Regression model. The dataset was divided into five folds, and the model was trained on four folds and validated on the remaining fold. This process was repeated five times, and the R² scores were averaged to obtain the mean cross-validation R².
```

```python
# Residual analysis

residuals = y_test - y_pred_final

plt.figure(figsize=(8, 6))

plt.scatter(y_pred_final, residuals)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Per Capita Water Use (Liters per Day)")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residual Analysis - Linear Regression")

plt.show()
#I performed residual analysis to examine the prediction errors of the final Linear Regression model. A residual is calculated as the actual value minus the predicted value. I plotted the residuals against the predicted Per Capita Water Use values. The horizontal zero line represents perfect prediction, and the distribution of residuals helps us identify whether the model errors have any systematic pattern.
#A residual is the difference between the actual value and the predicted value.
#A positive residual means the actual value is greater than the predicted value, so the model underestimated the value.
#A negative residual means the actual value is less than the predicted value, so the model overestimated the value.
```

```python
# Analyze Linear Regression coefficients

coefficient_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": linear_model_final.coef_
})

coefficient_df["Absolute Coefficient"] = (
    coefficient_df["Coefficient"].abs()
)

coefficient_df = coefficient_df.sort_values(
    by="Absolute Coefficient",
    ascending=False
)

print("Top 10 features by coefficient magnitude:")
display(coefficient_df.head(10))
#I analyzed the Linear Regression coefficients to understand the estimated contribution of each input feature to the model's predictions. Since the features were standardized, I compared the absolute coefficient magnitudes to identify the features with the strongest estimated linear effects. I displayed the top 10 features and retained the coefficient signs to understand whether their estimated effect was positive or negative.
#I used absolute values to compare the magnitude of the coefficients regardless of whether their effect was positive or negative.
#A negative coefficient means that, holding the other features constant, an increase in that standardized feature is associated with a decrease in the predicted Per Capita Water Use.
#Standardization puts the numerical features on a comparable scale, making coefficient magnitudes more meaningful when comparing their relative effects.
```
